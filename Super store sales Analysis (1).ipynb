```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("Train.csv")
df.head()
```

```python
print(df["Order Date"].head(10))
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    format="mixed",
    errors="coerce"
)

print(df["Order Date"].head(10))
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["Month"] = df["Order Date"].dt.month

monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
highest_month = monthly_sales.idxmax()
highest_sales = monthly_sales.max()

print("Highest Sales Month:", highest_month)
print("Highest Monthly Sales:", highest_sales)
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
best_region = region_sales.idxmax()
highest_region_sales = region_sales.max()

print("Best Performing Region:", best_region)
print("Highest Region Sales:", highest_region_sales)
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

print(segment_sales)
```

```python
segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
subcategory_sales = df.groupby("Sub-Category")["Sales"].sum()

print(subcategory_sales.sort_values(ascending=False))
```

```python
subcategory_sales.sort_values().plot(kind="barh", figsize=(10, 7))

plt.title("Sales by Sub-Category")
plt.xlabel("Total Sales")
plt.ylabel("Sub-Category")
plt.tight_layout()
plt.show()
```

```python
# Create all required sales calculations

category_sales = df.groupby("Category")["Sales"].sum()

region_sales = df.groupby("Region")["Sales"].sum()

segment_sales = df.groupby("Segment")["Sales"].sum()

product_sales = df.groupby("Product Name")["Sales"].sum()

subcategory_sales = df.groupby("Sub-Category")["Sales"].sum()


# Display Final Business Insights

print("===== SUPERSTORE SALES ANALYSIS =====")

print("Total Sales:", round(df["Sales"].sum(), 2))

print("Total Orders:", df["Order ID"].nunique())

print("Best-Selling Category:", category_sales.idxmax())

print("Best-Selling Region:", region_sales.idxmax())

print("Best Customer Segment:", segment_sales.idxmax())

print("Top-Selling Product:", product_sales.idxmax())

print("Top Sub-Category:", subcategory_sales.idxmax())
```
